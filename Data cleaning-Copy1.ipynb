```python
# Data load 

import pandas as pd 
df = pd.read_csv("sales_raw_new_2026.csv")
```

```python
df
```

```python
# duplicates remove

df = df.drop_duplicates()
```

```python
df
```

```python
#fix date column 

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```

```python
df = df.dropna(subset=['Order_Date'])
```

```python
df
```

```python
df['Profit']= df['Sales']-df['Cost']
df['Year']= df['Order_Date'].dt.year
df['Month']= df['Order_Date'].dt.month
```

```python
from sqlalchemy import create_engine

# Now create the engine
engine = create_engine('mysql+pymysql://root:root123@localhost:3306/salesdb')
```

```python
import pymysql

# Connect directly to your local MySQL server
conn = pymysql.connect(host='localhost', port=3306, user='root', password='root123')
cursor = conn.cursor()

# Command to physically create the database inside MySQL
cursor.execute("CREATE DATABASE IF NOT EXISTS salesdb;")
print("Database 'salesdb' created successfully! 🚀")

# Close connection
cursor.close()
conn.close()
```

```python
df.to_sql(name='sales_data',con=engine,if_exists='append',index= False)
```
